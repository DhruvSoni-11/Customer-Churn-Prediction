```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("cust_churn.csv")

df.head()
```

```python
df.info()
```

### Data Cleaning

```python
# drop customerID column
df.drop("customerID", axis=1, inplace=True)
```

```python
# convert 'TotalCharges' object -> float
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce') 
# convert NaN -> 0
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
# convert to catogarical -> numerical 
df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0})
df["Partner"] = df["Partner"].map({"Yes": 1, "No": 0})
df["Dependents"] = df["Dependents"].map({"Yes": 1, "No": 0})
```

```python
df["PhoneService"] = df["PhoneService"].map({"Yes": 1, "No": 0})
```

```python
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.show()
```

```python
pd.crosstab(df['Contract'], df['Churn'])
```

```python
df.groupby('Churn')['tenure'].mean()
```

Encoding

```python
df = pd.get_dummies(df, drop_first=True)
```

scaling

```python
# for scaling the numerical columns
scaler = StandardScaler()

nums_cols = ['tenure','MonthlyCharges','TotalCharges']

df[nums_cols] = scaler.fit_transform(df[nums_cols])
```

train | test | split

```python
from sklearn.model_selection import train_test_split

X = df.drop('Churn',axis=1)
Y = df['Churn']

# ratio of train/test:80/20
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.80, random_state=42)
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(Y_test, y_pred))
print(classification_report(Y_test, y_pred))
```

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt.fit(X_train, Y_train)

y_pred_dt = dt.predict(X_test)
```

Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()

rf.fit(X_train, Y_train)

y_pred_rf = rf.predict(X_test)
```

```python
# Accuracy:
print("Linear Regression Accuracy:", accuracy_score(Y_test, y_pred))
print("Decision Tree Accuracy:", accuracy_score(Y_test, y_pred_dt))
print("Random Forest Accuracy:", accuracy_score(Y_test, y_pred_rf))
```
